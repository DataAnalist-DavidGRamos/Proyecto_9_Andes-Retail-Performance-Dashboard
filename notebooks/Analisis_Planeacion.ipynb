#### 🖥️ Vista General (Overview)

**S (Situación):** Andes Retail Group ha consolidado $5.53 millones en ingresos con un margen robusto del 35.10% durante 2024-2025. Perú lidera con el 39% de participación ($2.16M) y simultáneamente muestra la mayor eficiencia operativa (35.17%), consolidándose como el motor de crecimiento rentable del grupo.

**C (Complicación):** Aunque las cuatro categorías mantienen márgenes homogéneos (34.88%-35.11%), el verdadero desafío radica en la volatilidad estacional. Deportes, siendo el mayor generador de volumen ($1.44M, 26%), concentra su demanda en periodos específicos, mientras que en Invierno la calidad del ticket cae 68% ($1,721 en Verano vs. $549 en Invierno), exponiendo la operación a ineficiencias logísticas.

**Q (Pregunta):** ¿Cómo mitigamos la volatilidad estacional para estabilizar márgenes sin sacrificar volumen?

**A (Respuesta):** Implementar estrategias de cross-selling dirigidas a estaciones bajas: (1) Bundling de Deportes con Electrónica en Invierno para restaurar ticket promedio, (2) Activar segmento Económico (9% del mix) mediante ofertas por mayor volumen en periodos críticos, y (3) Replicar dinámicas de venta del trimestre Verano (donde Premium domina con 47% de ingresos) en Primavera para crear un "secundario" de demanda.

#### 🔎 Vista Detalle - Análisis Estacional y de Segmentación

**S (Situación):** Verano consolida $2.24 millones (40.5% del ingreso anual) en 1,302 pedidos con un ticket promedio de $1,721.75. El segmento Premium concentra el 47% de los ingresos del grupo, alcanzando su pico máximo durante este trimestre, demostrando una demanda premium acumulada.

**C (Complicación):** Invierno sufre una contracción severa: ingresos caen a $653K (-71% vs. Verano), con solo 1,190 pedidos pero tickets colapsados a $549.17 (-68% en valor unitario). El problema no es menos clientes, sino clientes gastando significativamente menos. El 82.5% de transacciones invernales quedan por debajo de la mediana de mercado ($896), saturando la operación con pedidos que apenas cubren costos logísticos.

**Q (Pregunta):** ¿Esta caída es comportamiento estacional inevitable o refleja falla en la estrategia de posicionamiento invernal?

**A (Respuesta):** El análisis del mix de productos por estación sugiere que Invierno concentra compras de Ropa y Hogar (categorías de bajo ticket: $1.32M y $1.36M respectivamente), mientras Verano favorece Deportes y Electrónica (tickets mayores). 

Acciones recomendadas:
1. **Bundles Estratégicos:** Combinar Electrónica + Ropa en Invierno para restaurar ticket hacia $1,100+ (impacto estimado: +$400K en ingresos, +$140K en ganancia)
2. **Segmentación Premium:** Crear ofertas exclusivas para Premium en Invierno (47% del ingreso grupo pero solo 40% del ticket invernal)
3. **Promoción Cruzada:** Implementar "Complemento Gratis" en Deportes cuando compren Ropa, aumentando el artículo promedio (AOV)

**Análisis Adicional: Segmentación de Clientes**

**S (Situación):** El segmento Premium representa 1,519 clientes (30.4% del total) que generan $2.60 millones (47% del ingreso grupal) con un margen de 35.14%. Estándar aporta 2,504 clientes (50.1%) con $2.44M (44% del ingreso) y margen de 34.94%. En contraste, Económico tiene 977 clientes (19.5%) pero genera solo $497K (9%), a pesar de mantener un margen competitivo de 35.06%.

**C (Complicación):** El segmento Económico es una "mina sin explotar": representa el 19.5% de la base de clientes pero aporta solo el 9% de ingresos, subutilizando oportunidades de monetización. Esto sugiere que nuestras estrategias de "upsell" desde Económico a Estándar son inefectivas.

**Q (Pregunta):** ¿Por qué casi 1 de cada 5 clientes en Económico no migra a categorías de mayor valor, a pesar de márgenes competitivos?

**A (Respuesta):** El análisis por estación revela que Económico mantiene compras pequeñas incluso en Verano ($785 vs. $2,714 en Premium), sugiriendo una estratificación de demanda persistente. Recomendaciones:
1. **Capas de Bundling Económico:** Crear "Combo Premium" (2 productos = descuento 5%) para elevar AOV de $497 a $650 estimado (+31% en ingresos Económico)
2. **Inversión en Reactivación:** Implementar programa de loyalty para Premium (descuentos por volumen), consolidando la retención del 47% de ingresos
3. **Estrategia de Transición:** Usar datos de comportamiento en Verano (cuando es más probable que Económico compre Premium) para activar campañas de upsell estacionales
