#### Vista Overview - ¿Cómo ha evolucionado el ingreso total entre 2024 y 2025?

**La pregunta que realmente responderé será:**
¿Estamos creciendo en los lugares y con los clientes adecuados, o ese aumento en ingresos oculta que ganamos menos por cada venta?

Especifica qué KPIs incluirás y por qué: (3-4 KPIs)
- **KPI 1: Ingresos Totales.** *¿Cómo evoluciona nuestra escala de facturación continua en el tiempo?* (Mide el ritmo de crecimiento del mercado bruto).
- **KPI 2: Margen % (Promedio).** *¿Qué región (geografía) lidera verdaderamente la eficiencia operativa?* (Permite comparar países de distinto tamaño justa y equitativamente).
- **KPI 3: Ganancia Total.** *¿Qué categorías de productos nos hacen retener más utilidades?* (Identifica qué parte del mix de productos financia a la empresa).
- **KPI 4: Ticket Promedio.** *¿Nuestras ventas dependen de transacciones masivas pequeñas o de ventas premium?* (Mide la calidad del flujo de caja).

Explica qué gráfico usarás para cada pregunta y justifica la elección.

- **Gráfico de tiempo:** Gráfico de líneas de `Ingresos` vs `Ganancia` por `Mes`. *Justificación:* Título sugerido: "Los ingresos crecen - pero el margen no los sigue". Las líneas son el único formato que evidencia tendencias continuas; la tensión narrativa forzará al directivo a buscar variaciones temporales.
  
- **Gráfico comparación geográfica:** Gráfico de **Barras horizontales** de `Ingresos` por `País` con **indicador visual (línea de Margen %)**. *Justificación:* Facilita la comparación direccional de escala, pero al mostrar el margen simultáneamente evita la falsa conclusión de que "el país que más vende es siempre el mejor".

- **Gráfico comparación clientes/productos:** **Gráfico de Columnas y Líneas** - Columnas agrupadas de Ingresos con Línea de Margen % por `Categoría_Producto`. *Justificación:* Al tener volúmenes de ingresos muy parecidos entre categorías, las barras permiten notar la sutil diferencia de escala de forma más precisa que un gráfico de área. La inclusión de la línea de Margen % revela inmediatamente si la categoría con más ventas es también la más eficiente.

- **Gráfico de Anillos:** de `Ingresos` por `Nivel_Venta`. *Justificación:* Visualiza directamente y de golpe la distribución entre clientes de bajo valor frente a clientes de ticket premium.

Descripción de la jerarquía visual:
- **Superior Izquierdo:** Título narrativo y Filtro de Año.
- **Fila superior:** 4 Tarjetas KPI simétricas (Ingresos, Ganancia, Margen, Ticket).
- **Centro:** Gráfico de evolución temporal (el de mayor tamaño).
- **Base:** Gráficos de desglose regional y por producto debajo de la línea temporal.

Filtro(s)
- **Año:** Para contrastar el desempeño anual 2024 vs 2025.
- **País:** Para focalizar el análisis regional.



#### Vista detalle - ¿Por qué se han caído/subido/mantenido los ingresos?

**Pregunta Global:** ¿Qué factores de causalidad (estacionalidad y segmentación) explican las variaciones registradas?

Explica qué gráfico usarás para cada pregunta y justifica la elección.

- **Gráfico comparación estacional contra ingresos:** Gráfico de **Líneas dobles con eje secundario** - Eje Y1: `Ingresos`; Eje Y2: `Margen %` por segmento temporal de `Estación`. *Justificación:* Permite ver la "divergencia" exacta. Si la línea de Ingresos sube en Invierno pero la de Margen baja, el gráfico detecta el "crecimiento hueco" inmediato.
  
- **Gráfico comparación Calidad de Venta contra ingresos:** Gráfico de **Barras 100% apiladas** - `Estación` en eje X y proporción de `Nivel_Venta` en Y. *Justificación:* No basta con saber qué vende cada estación; necesitamos responder *cuándo cambió* la composición de los tickets altos vs bajos.

- **Gráfico comparación Rentabilidad contra Segmento:** Gráfico de **Barras simples** - `Segmento_Cliente` en eje Y vertical y `Margen %` en eje X. *Justificación:* Una comparación limpia que responde rápidamente a la hipótesis de eficiencia sin contaminar visualmente unidades en dinero ($) con porcentajes (%).



- Tabla mostrando 4-6 columnas:
  - Conteo `ID_Pedido` (Métrica de validación de volumen operativo).
  - Conteo único `ID_Cliente` (Valida dependencia y fidelidad de base).
  - `Región` (Perspectiva geográfica sin duplicar granularidad de país).
  - `Ingresos` (Total monetario de la muestra aislada).
  - `Margen %` (Alertas condicionales verde/rojo de rentabilidad).
  - `Nivel_Venta` (Contexto de nivel comercial de la transacción).
 
Filtros (mínimo 2)
- `Fecha_pedido` (Presets de filtrado veloz ejecutivo).
- `Segmento_Cliente` (Filtro central para segmentación).
- `Categoría_Producto` (Vital para aislar y desmentir qué productos impactan la estacionalidad).
- `País` (Dimensionado regional directo sin necesidad de una gráfica extra).

#### 🖥️ Vista General (Overview)

**S (Situación):**  Andes Retail Group ha consolidado unos ingresos de $5.53 millones con un margen de rentabilidad sumamente estable del 35.10% durante el periodo 2024-2025. Perú se posiciona como el mercado líder con el 39% de la participación total.

**C (Complicación):**  Aunque el crecimiento es constante, la categoría de mayor volumen (Deportes) es la que presenta el margen de eficiencia más bajo, mientras que Electrónica lidera en creación de valor.

**Q (Pregunta):**   ¿Cómo podemos asegurar que el crecimiento en volumen no erosione la rentabilidad global del grupo?

**A (Respuesta):**  Es imperativo replicar las estrategias de precios de Electrónica en Deportes y fortalecer la operación en regiones premium de Chile que muestran márgenes superiores al promedio.

---

#### 🔎 Vista Detalle

**S (Situación):**  El análisis estacional muestra un ecosistema saludable durante el Verano, donde las "Ventas Altas" dominan el 89.63% de la facturación.

**C (Complicación):**  En Invierno, la calidad de la venta sufre un deterioro drástico: el 70.68% de los pedidos caen en la categoría de "Venta Baja", lo que estresa los costos operativos por pedido.

**Q (Pregunta):**  ¿Qué factor está impulsando la caída del valor del ticket durante la temporada invernal?

**A (Respuesta):**  Los datos sugieren una dependencia de productos de bajo costo en el cierre de año. Se recomienda implementar estrategias de "Bundling" (Venta cruzada) en Invierno para elevar el ticket promedio y diluir los costos logísticos.
